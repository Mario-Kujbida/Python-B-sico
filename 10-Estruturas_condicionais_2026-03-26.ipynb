## 01 – Dados as operações abaixo, em Python, qual o resultado da operação lógica: 
p and q and r and s and t?
<br>
<br>Imprima os objetos e o resultado usando uma estrutura condicional.

<br>p = 2.0 + 3.0 / 5.0 == 10
<br>q = (p + 10.2) - 5.0 * 2.0 <= 2.9
<br>r = 10 <= p / q + 33.8 < 100

```python
p = 2.0 + 3.0 / 5.0 == 10
q = (p + 10.2) - 5.0 * 2.0 <= 2.9
r = 10 <= p / q + 33.8 < 100
s = 5 != p / (q + r)
t = 33 == s * 2

print("p:", p)
print("q:", q)
print("r:", r)
print("s:", s)
print("t:", t)

resultado = p and q and r and s and t

print("Resultado:", resultado)

if resultado:
    print("Todos são verdadeiros")
else:
    print("Existe pelo menos um valor falso")
```

## Instancie um objeto da classe array numpy, contendo 5 valores quantitativos randômicos normalmente distribuídos, com média 10 e desvio padrão 3. 
Quantos e quais são os valores superiores a 10?

```python
import numpy as np

array = np.random.normal(loc=10, scale=3, size=5)

print("Array:", array)
```

```python
# Etapa 1: gerar dados, que está logo abaixo da questão, o array.

# Etapa 2: identificar valores maiores que 10
maiores_10 = array[array > 10]

# Etapa 3: contagem
quantidade = len(maiores_10)

# Etapa 4: saída
print("Array gerado:", array)
print("Valores maiores que 10:", maiores_10)
print("Quantidade de valores maiores que 10:", quantidade)

# Estrutura condicional
if quantidade > 0:
    print("Existem valores maiores que 10.")
else:
    print("Não existem valores maiores que 10.")
```

## No exercício 7 da aula 6, foi instanciado um objeto da classe DataFrame. Exporte (usando nome_data_frame.to_csv(‘nome_arquivo.csv’)) para .csv estes dados e os importe (com pd.read_csv(‘nome_arquivo.csv’) aqui. Edite o objeto para que os index sejam os nomes dos genes e para que as colunas sejam facilmente identificadas. Usando o número de caracteres do primeiro gene, quero saber a informação gênica. Caso o nome tenha entre 2 e 4 caracteres inclusive, posso saber as informações do gene? Porém, caso tenha menos de 2 caracteres, posso saber todos os valores nas instâncias onde os números de caracteres dos nomes são superiores a 4? Ainda, caso tenha mais de 4 caracteres, apenas me informe quem é superior.


```python
import pandas as pd
import numpy as np
```

```python
df_novo = pd.read_csv('exercicio_10.csv', index_col='Unnamed: 0').T
df_novo
```

## 07 – Descreva o racional da sua resposta:

R.: O problema foi resolvido a partir da análise do número de caracteres do primeiro gene, utilizando a informação já presente na coluna “Caracteres” do DataFrame, evitando recalcular os valores.

A lógica consiste em:

Identificar o número de caracteres do primeiro gene.
Utilizar estruturas condicionais (if, elif) para decidir qual ação executar.
Aplicar filtros no DataFrame com base no tamanho dos nomes dos genes.

As condições foram definidas a partir:

Se o número de caracteres estiver entre 2 e 4 (inclusive), é possível acessar diretamente as informações do gene.
Se for menor que 2, são selecionados apenas os genes com nomes superiores a 4 caracteres.
Se for maior que 4, também são selecionados os genes com nomes superiores a 4 caracteres.

A filtragem é feita utilizando o pandas.

## 09 – Escreva o programa em Python:

```python
import pandas as pd

# Importar o CSV

# Garantir que a coluna "Caracteres" é numérica
df_novo['Caracteres'] = pd.to_numeric(df_novo['Caracteres'])

# Selecionar o primeiro gene
primeiro_gene = df_novo.index[0]

# Número de caracteres (usando a coluna já existente)
n_caracteres = df_novo.iloc[0]['Caracteres']

# Vetor com tamanhos dos nomes
tamanhos = df_novo['Caracteres']

# Estrutura condicional
if 2 <= n_caracteres <= 4:
    print("Informações do primeiro gene:")
    print(df_novo.loc[primeiro_gene])

elif n_caracteres < 2:
    print("Genes com mais de 4 caracteres:")
    print(df_novo[tamanhos > 4])

elif n_caracteres > 4:
    print("Genes com mais de 4 caracteres:")
    print(df_novo[tamanhos > 4])
```
