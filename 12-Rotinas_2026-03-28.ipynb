Instancie um objeto da classe NumPy array, contendo 5 valores quantitativos randômicos normalmente distribuídos, com média 10 e desvio padrão 3.<br><br>
Quantos e quais são os valores superiores a 10?

## 02 – Descreva o racional da sua resposta:

R.: Inicialmente, é gerado um array com valores aleatórios seguindo distribuição normal (média 10 e desvio padrão 3). Em seguida, cria-se o objeto “maiores_10”, que filtra os valores superiores a 10. Depois, o objeto “quantidade” calcula quantos valores atendem a esse critério, e por fim os resultados são exibidos.

## 04 – Escreva o programa em Python:

```python
import numpy as np

def analisar_valores():
    # Etapa 1: gerar o array
    array = np.random.normal(loc=10, scale=3, size=5)
    
    # Etapa 2: identificar valores maiores que 10
    maiores_10 = array[array > 10]
    
    # Etapa 3: contagem
    quantidade = len(maiores_10)
    
    # Etapa 4: saída
    print("Array gerado:", array)
    print("Valores maiores que 10:", maiores_10)
    print("Quantidade de valores maiores que 10:", quantidade)
    
    # Estrutura condicional
    if quantidade > 0:
        print("Existem valores maiores que 10.")
    else:
        print("Não existem valores maiores que 10.")

# Chamando a função
analisar_valores()
```

## 05 – Seria possível, com o racional proposto, saber quantos e quais valores são superiores a 20? Porque?

R.: Sim, seria possível, pois o racional utiliza uma condição de filtragem. Bastaria alterar a condição de > 10 para > 20, mantendo a mesma lógica, para obter tanto os valores quanto a quantidade acima de 20.

```python
def analisar_valores():
    # Etapa 1: pegar o array
    array_2 = np.random.normal(loc=10, scale=3, size=5)

    # Etapa 2: identificar valores maiores que 10
    maiores_20 = array_2[array_2 > 20]
    
    # Etapa 3: contagem
    quantidade_2 = len(maiores_20)
    
    # Etapa 4: saída
    print("Array gerado:", array_2)
    print("Valores maiores que 20:", maiores_20)
    print("Quantidade de valores maiores que 20:", quantidade_2)
    
    # Estrutura condicional
    if quantidade_2 > 0:
        print("Existem valores maiores que 20.")
    else:
        print("Não existem valores maiores que 20.")

# Chamando a função
analisar_valores()
```

## Desenvolva um programa de computador que retorne a média de um conjunto de valores.

## 06 – Descreva o racional da sua resposta:

R.: consiste em receber um conjunto de valores, somar todos os elementos e dividir pelo total de valores, obtendo assim a média. Essa lógica é expressa em uma função, que recebe os dados como entrada e retorna o resultado, garantindo organização e reutilização do código.

## 07 – Escreva o programa em Python:

```python
def calcular_media(valores):
    soma = sum(valores)
    quantidade = len(valores)
    media = soma / quantidade
    return media

# Gerando os dados
valores = np.random.randint(0, 100, 7).tolist()
print("Valores:", valores)

# Calculando a média
media = calcular_media(valores)
print("Média:", media)
```

## 09 – Desenvolva um programa de computador que retorne a divisão de dois objetos simples. Seu programa deve ser capaz de reconhecer a classe destes objetos e o valor, antes de retornar ao usuário.

```python
def dividir(a, b):
    print("Objeto A:", a, "| Classe:", type(a))
    print("Objeto B:", b, "| Classe:", type(b))
    
    # Verificando se são números
    if isinstance(a, (int, float)) and isinstance(b, (int, float)):
        if b != 0:
            resultado = a / b
            return resultado
        else:
            return "Erro: divisão por zero."
    else:
        return "Erro: os objetos não são numéricos."

# Exemplo de uso
x = 10
y = 2.5

print("Resultado da divisão:", dividir(x, y))
```

## 10 – Desenvolva um programa de computador que retorne a divisão de dois objetos compostos. Seu programa deve ser capaz de reconhecer a classe destes objetos e o valor, antes de retornar ao usuário.

```python
def div_listas(c, d):
    print("c:", c, "| classe:", type(c))
    print("d:", d, "| classe:", type(d))

    # Verifica se são listas
    if not isinstance(c, list) or not isinstance(d, list):
        return "Erro: os objetos não são listas."

    # Verifica tamanho
    if len(c) != len(d):
        return "Erro: listas com tamanhos diferentes."

    resultado = []
    
    for i in range(len(c)):
        if isinstance(c[i], (int, float)) and isinstance(d[i], (int, float)):
            if d[i] != 0:
                resultado.append(c[i] / d[i])
            else:
                return "Erro: divisão por zero."
        else:
            return "Erro: elementos não numéricos."

    print("resultado:", resultado, "| classe:", type(resultado))
    return resultado


lista1 = [2, 6]
lista2 = [30, 45, 80]

print(div_listas(lista1, lista2))
```
