## 10 – Importe o data.csv (dados de seguro na Flórida em 2012) para uma DataFrame. Armazene em um dicionário o número de entradas, uma lista com as classes das colunas e outra lista com o número de elementos de cada classe, três listas, contendo a média, o desvio padrão e o terceiro quartil da primeira, quarta e nona colunas . Recupere o dicionário.

```python
df_Flórida2012 = pd.read_csv('data.csv')
df_Flórida2012
```

```python
entradas = df_Flórida2012.shape
entradas
```

```python
classes = list(df_Flórida2012.dtypes.unique())
classes
```

```python
n_classes = df_Flórida2012.dtypes.value_counts()
n_classes
```

```python
colunas = df_Flórida2012.iloc [: , [0, 3, 8]]
colunas
```

```python
media = colunas.mean()
media
```

```python
desv_pad = colunas.std()
desv_pad
```

```python
terc_quartil = colunas.quantile(0.75)
terc_quartil
```

```python
dicionario_Florida2012 = {
    'N_ENTRADAS': entradas,
    'CLASSES': classes,  
    'N_CLASSES': n_classes,
    'MÉDIAS': media,
    'D_PAD': desv_pad, 
    '3Q': terc_quartil 
}
dicionario_Florida2012
```
