## Instancie um objeto da classe lista contendo 5.000 valores inteiros iniciando em 0. 
Quantos são pares e quantos são impares?

## 01 – Descreva o racional da sua resposta:

R.: Inicialmente é criada uma lista com 5.000 números inteiros consecutivos, de 0 a 4999. Para descobrir se os números são pares é necessário dividir os números por 2 e o resultado da divisão ser igual a 0, enquanto ímpares não possuem uma divisão inteira e exata. Como a sequência é contínua e inicia em 0 (par), há alternância entre pares e ímpares. Assim, a distribuição é uniforme, resultando em 2.500 números pares e 2.500 números ímpares.

## 02 – Desenhe o fluxograma da resposta:

## 03 – Escreva o programa em Python:

```python
# Criar lista de 0 até 4999
lista = list(range(5000))

# Contadores
pares = 0
impares = 0

# Percorrer a lista
for numero in lista:
    if numero % 2 == 0:
        pares += 1
    else:
        impares += 1

# Resultado
print("Quantidade de pares:", pares)
print("Quantidade de ímpares:", impares)
```

## 04 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
total = 5000

pares = total // 2
impares = total // 2

print("Pares:", pares)
print("Ímpares:", impares)
```
