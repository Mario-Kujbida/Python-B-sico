## Importe o .csv utilizado no exercício 7 da aula 11 e instancie um objeto da classe DataFrame. Considere que há uma ordem entre os genes.
Como inserir  na DataFrame uma tabela de frequência do número de caracteres no nome dos genes? Uma vez inserida, como obter um gráfico de barras deste dado?

## 04 – Descreva o racional da sua resposta:

R.: Primeiramente o arquivo .csv é importado e convertido em um DataFrame, preservando a ordem original dos genes. Em seguida, calcula-se a frequência do número de caracteres utilizando uma função. Essa informação é então associada ao DataFrame por meio de mapeamento, criando uma nova coluna “Frequencia” sem alterar a ordem dos dados. Por fim, utiliza-se essa distribuição para gerar um gráfico de barras, permitindo a visualização das frequências.

## 05 – Desenhe o fluxograma da resposta:

## 06 – Escreva o programa em Python:

```python
# Importar o CSV
df_novo = pd.read_csv('exercicio_10.csv', index_col='Unnamed: 0').T
print(df_novo)

# Garantir que a coluna é numérica
df_novo["Caracteres"] = pd.to_numeric(df_novo["Caracteres"])

# Calcular frequência
freq = df_novo["Caracteres"].value_counts()

# Adicionar coluna "Frequencia" mantendo a ordem
df_novo["Frequencia"] = df_novo["Caracteres"].map(freq)

print(df_novo)

# Gráfico de barras (distribuição)
freq.sort_index().plot(kind="bar")

plt.xlabel("Número de caracteres")
plt.ylabel("Frequência")
plt.title("Distribuição do número de caracteres dos genes")

plt.show()
```
