No paradigma de Programação Orientada a Objetos, o que é um Objeto?
<br>
<br>
R.: Na POO objeto é toda instância de uma classe e variáveis são objetos instanciados, ou seja, todo objeto tem inicio na sua classe, que seria como você vai fazer para chegar ao objeto, enquanto a instância é como você vai fazer, o passo a passo e técnica para chegar até o objeto final.

```python
a = 13
a.count ("1")
```

```python
object_str = "Mario"
object_str.count("a")
```
