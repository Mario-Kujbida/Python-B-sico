```python
def contar_inteiros (object: list | tuple) -> int:
    contador = 0
    for x in range(len(object)):
        if type(object[x]) == int:
            contador += 1
    print(f'Quantidade de inteiros: {contador}')
    return contador
```

```python
object = [1, 2, 3, 'a', 'b', 'c', 4.5, 5.6, 6.7]
resultado_contador = contar_inteiros(object=object)
print(resultado_contador)
```
